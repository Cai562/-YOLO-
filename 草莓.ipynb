```python
import os
import yaml
from ultralytics import YOLO

def main():
    # ==========================================
    # 1. 路径设置 (强行绑定绝对路径)
    # ==========================================
    dataset_root = "E:/w-output/Strawberry"
    yaml_path = f"{dataset_root}/data.yaml"
    
    print("正在优化并修复 data.yaml 中的相对路径...")
    with open(yaml_path, 'r', encoding='utf-8') as f:
        data = yaml.safe_load(f)
    
    data['path'] = dataset_root
    data['train'] = "train/images"
    data['val'] = "valid/images"
    data['test'] = "test/images"
    
    fixed_yaml_path = f"{dataset_root}/data_fixed.yaml"
    with open(fixed_yaml_path, 'w', encoding='utf-8') as f:
        yaml.safe_dump(data, f, allow_unicode=True)

    # ==========================================
    # 2. 加载模型
    # ==========================================
    weights_name = "yolov8n.pt"
    print("正在加载轻量化 YOLOv8 Nano 模型...")
    model = YOLO(weights_name) 
    
    # ==========================================
    # 3. 开始防崩溃安全训练配置
    # ==========================================
    print("开始训练草莓成熟度检测网络 (安全防崩溃模式)...")
    results = model.train(
        data=fixed_yaml_path,   
        epochs=50,             
        batch=4,               # 💡 降低批次大小：从 16 降到 4，大幅减少内存和 CPU 瞬时负载
        imgsz=320,             # 💡 降低分辨率：从 640 降到 320，极大地减轻 CPU 计算压力，防止溢出
        device='cpu',          
        workers=0,             # 💡 彻底禁用多线程：单线程读取图片，根治 Windows 多进程死锁造成的崩溃
        amp=False,             # 💡 禁用混合精度：纯 CPU 训练必须关闭 amp，否则极易引发内核底层的 C++ 崩溃
        cache=False,           # 💡 禁用内存缓存：不把图片全塞进内存，随用随读，保护你的运行内存
        plots=True             
    )
    
    print("🎉 训练完成！")

if __name__ == "__main__":
    main()
```
